Dataset generation 

```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_blobs, make_moons, make_circles


# K-Means Dataset


X, y = make_blobs(
    n_samples=400,
    centers=4,
    cluster_std=1.2,
    random_state=42
)

df = pd.DataFrame(X, columns=["x", "y"])
df.to_csv("kmeans_dataset.csv", index=False)



# K-Medoids Dataset (with outliers)

X, y = make_blobs(
    n_samples=300,
    centers=3,
    random_state=42
)

outliers = np.random.uniform(-10, 10, (30, 2))
X = np.vstack((X, outliers))

df = pd.DataFrame(X, columns=["x", "y"])
df.to_csv("kmedoids_dataset.csv", index=False)


# Agglomerative Dataset


X, y = make_blobs(
    n_samples=300,
    centers=3,
    random_state=42
)

df = pd.DataFrame(X, columns=["x", "y"])
df.to_csv("agglomerative_dataset.csv", index=False)



# Divisive Dataset

X, y = make_blobs(
    n_samples=400,
    centers=4,
    random_state=42
)

df = pd.DataFrame(X, columns=["x", "y"])
df.to_csv("divisive_dataset.csv", index=False)



# DBSCAN Dataset (Moons)


X, y = make_moons(
    n_samples=400,
    noise=0.05,
    random_state=42
)

df = pd.DataFrame(X, columns=["x", "y"])
df.to_csv("dbscan_dataset.csv", index=False)



# Extra Dataset (Circles)


X, y = make_circles(
    n_samples=400,
    noise=0.05,
    factor=0.5,
    random_state=42
)

df = pd.DataFrame(X, columns=["x", "y"])
df.to_csv("circles_dataset.csv", index=False)


print("All datasets generated successfully!")
```

K-Means Clustering

```python
import os
os.environ["OMP_NUM_THREADS"] = "2"

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

data = pd.read_csv("kmeans_dataset.csv")

X = data[['x','y']].values

model = KMeans(n_clusters=4, random_state=42)

labels = model.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=labels)
plt.scatter(model.cluster_centers_[:,0], model.cluster_centers_[:,1], marker='X', s=200)

plt.title("KMeans Clustering")
plt.show()
```

K-Medoids Clustering

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn_extra.cluster import KMedoids

data = pd.read_csv("kmedoids_dataset.csv")

X = data[['x','y']].values

model = KMedoids(n_clusters=3, random_state=42)

labels = model.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=labels)

centers = model.cluster_centers_
plt.scatter(centers[:,0], centers[:,1], marker='X', s=200)

plt.title("K-Medoids Clustering")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering

data = pd.read_csv("agglomerative_dataset.csv")

X = data[['x','y']].values

model = AgglomerativeClustering(n_clusters=3)

labels = model.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=labels)

plt.title("Agglomerative Clustering")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

Divisive Clustering

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import BisectingKMeans

data = pd.read_csv("divisive_dataset.csv")

X = data[['x','y']].values

model = BisectingKMeans(n_clusters=4, random_state=42)

labels = model.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=labels)

plt.title("Divisive Clustering")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN

data = pd.read_csv("dbscan_dataset.csv")

X = data[['x','y']].values

model = DBSCAN(eps=0.2, min_samples=5)

labels = model.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=labels)

plt.title("DBSCAN Clustering")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
